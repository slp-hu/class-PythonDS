## 3. 集計用乗率（Weight）― 1人が「何人分」を代表するか 🖐

ここが、教科書データと実データのいちばん大きな違いです。じっくりいきます。

**なぜ「行数＝人口」ではないのか。**
前節のとおり、これは**一部の人だけを選んで調べた標本調査**です。たとえば「ある地域・ある属性の人を、500人に1人くらいの割合で選んで調べた」としましょう。すると、調査に選ばれた**1人**は、自分と似ていて**調べられなかった約500人の分まで代表している**ことになります。

この「**1人が母集団の何人分にあたるか**」という倍率が、**集計用乗率（しゅうけいようじょうりつ）＝ Weight**です（「拡大乗率」とも呼びます）。

- だから人口を出すときは、**行（調査票）の枚数を数える**のではなく、**一人ひとりの Weight（代表する人数）を全部足し合わせます**。
- このデータでは Weight は平均しておよそ **500**（最小200〜最大1500）。ざっくり **1行 ≈ 約500人分**、というイメージです。

整理すると：

| 何を計算するか | 意味 | このデータでの値 |
|---|---|---|
| **行数を数える** | **調べた記録の数（標本の大きさ）** | 220,391 |
| **Weight を合計する** | **それらが代表する母集団の人数** | 110,194,700（＝全国の15歳以上人口） |

直感的には **行数 × 平均Weight ≈ 母集団** です。実際 `220,391 × 約500 ≈ 1.1億` になります。
だから「行数（22万）」を人口だと思って報告すると、本当の人口（約1.1億）の **約1/500** しか数えていないことになります。これが「**行を数えてはいけない**」の意味です。

**つまずきやすい点（読んでおくと安心）：**
- **Weight が小数になるのはなぜ？** Weight は「人数」そのものではなく「倍率（選ばれた割合の逆数）」だからです。たとえば `504.5` のような小数になります。小数のまま足すからこそ、合計が母集団の人数にぴったり合います。
- **なぜ人によって Weight が違う（200〜1500）？** グループごとに選ばれる割合が違うからです。あまり選ばれなかったグループの1人は、より多くの人を代表する（＝Weight が大きい）。だから「全員に同じ数を掛ける」のではなく、**一人ひとりの Weight で重みづけして足す**必要があります。

では、まず行数と Weight 合計がまったく違うことを、自分の手で確かめましょう。


**【やってみよう 3】** 行数と Weight 合計を出して比べよう。`Weight` は文字列なので、まず数値に直す必要がある。

> ✅ **自己チェック**：Weight 合計 = **110,194,700**（＝全国の15歳以上人口）。行数（22万）とまったく違うはず。

ここで使う `pd.crosstab` は教科書には出てこないので、先に説明します。

**クロス集計**とは、**2つの分類軸を掛け合わせて、組合せごとの件数を数えた表**のことです。
- タテ（行）に一方の分類、ヨコ（列）にもう一方の分類を置き、
- 交わる**マス目**に「その組合せに当てはまる行（記録）が何件あるか」を入れます。

書き方はこうです：

```python
pd.crosstab(タテに並べたい列, ヨコに並べたい列)
```

今回は **タテ＝有業／無業（`T_WorkStatus`）**、**ヨコ＝配偶関係（`T_NoworkMarriage`）** を掛け合わせます。
すると「有業者で配偶者あり」「無業者で未婚」…といった**各組合せの件数**が一覧になります。

**`.fillna('(非該当)')` を付ける理由**：配偶関係が空欄（NaN）の行は、そのままだと集計表に列として現れず「消えて」しまいます。`(非該当)` という見出しに置き換えると、**空欄がどこにどれだけあるか**も表の中で見えるようになります。

**読み方のコツ**：表の**どのマスに件数が集まっているか**を見ます。
たとえば「有業者」の行で `(非該当)` の列にだけ件数が集まり、未婚・配偶者あり・死別離別の列がほぼ 0 なら、
それは「**配偶関係は有業者には記録されていない（無業者にだけある）**」という意味になります（本当にそうなるか、下で確かめましょう）。

> ※ ここでは件数（行数）を数えています。人口として正確に出すなら本来は Weight で重みづけしますが、
> いまの目的は「**配偶関係がどの人たちに記録されているか（＝列の適用範囲）**」を確かめることなので、件数で十分です。


````python
pd.crosstab(df['T_WorkStatus'], df[____].fillna('(非該当)'))
````

## 7. 機会格差を可視化する（その1）― 男女の正規雇用率 🖐

ここからが個票の威力。**正規雇用率＝正規 ÷（正規＋非正規）** を男女別に自分で計算します。
母数は「職員・従業員」だけ（非該当は除く）。


**【やってみよう 7】** 男女別の正規雇用率を計算して棒グラフにしよう。

> ✅ **自己チェック**：男 ≈ **77.8%** / 女 ≈ **46.8%**。


````python
# 母数＝職員・従業員だけ（emp は §6 で作成済み。作り直してもよい）
emp = df[df['T_WorkRegular'].isin(['正規の職員・従業員','非正規の職員・従業員'])].copy()
# 「正規なら Weight、そうでなければ 0」の列を作る
emp['正規Weight'] = (emp['T_WorkRegular']==____) * emp['Weight']

rate_g = emp.groupby('T_Gender').apply(
    lambda x: x['正規Weight'].sum()/x['Weight'].____()*100, include_groups=False)
print(rate_g.round(1))

ax = rate_g.reindex(['男','女']).plot.bar(rot=0, color=['#4C72B0','#C44E52'])
ax.set_ylabel('正規雇用率 (%)'); ax.set_title('男女別 正規雇用率（全国）')
for i,v in enumerate(rate_g.reindex(['男','女'])): ax.text(i, v+1, f'{v:.1f}%', ha='center')
plt.show()
````

男女差がはっきり出ます。ただし「**これは結論か、仮説か**」を問うのは次回（第12回）。
今日の成果は、**個票から自分の手で“格差の数字”を作れた**ことです。

````python
rate_pref = (emp.groupby(['T_Prefecture','T_Gender'])
             .apply(lambda x: x['正規Weight'].sum()/x['Weight'].sum()*100,
                    include_groups=False).____())   # 男女を列にするメソッドは？
josei = rate_pref['女'].sort_values()
print('最低:', josei.index[0], round(josei.iloc[0],1), '/ 最高:', josei.index[-1], round(josei.iloc[-1],1))
ax = josei.plot.barh(figsize=(6,9), color='#C44E52')
ax.set_xlabel('女性の正規雇用率 (%)'); ax.set_title('都道府県別 女性の正規雇用率（低い順）')
plt.tight_layout(); plt.show()
````
